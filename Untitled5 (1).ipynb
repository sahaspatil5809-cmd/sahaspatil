```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
for col in cat.columns:
    cat[col] = le.fit_transform(cat[col])

display(cat.head())
```

```python
# Concatenate the numerical and encoded categorical dataframes
df = pd.concat([num, cat], axis=1)

# Display the first few rows of the final dataframe
display(df.head())
```

```python
from sklearn.model_selection import train_test_split
x = df.drop('Salary', axis=1)
y = df['Salary']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

# Before initializing the model, we need to handle NaN values in our data.
# Let's inspect the x_train to confirm NaNs and then address them.
# A common approach is to drop rows with NaN values if they are few,
# or to impute them. For now, let's check for NaNs in x_train.
# Since the error occurred during model.fit(x_train, y_train),
# the NaNs must be in x_train.

# One way to handle this is to drop NaNs from the entire DataFrame (df)
# before splitting, then re-split and train.
# However, to isolate the fix to this cell or immediately preceding it,
# we will add a check and a simple dropna operation on x_train and y_train.

# You should ideally handle missing values earlier in your preprocessing pipeline,
# potentially before the train-test split, to ensure consistency.

# Let's check for NaNs in x_train and y_train again before fitting
# print("NaNs in x_train:\n", x_train.isnull().sum()[x_train.isnull().sum() > 0])
# print("NaNs in y_train:\n", y_train.isnull().sum())

# For demonstration, let's remove rows with any NaN values from x_train and corresponding y_train
# Note: This is a simplistic approach; for production, consider imputation or more refined strategies.

# Create a combined DataFrame for easy dropping of corresponding rows
combined_train_df = pd.concat([x_train, y_train], axis=1)
original_rows = len(combined_train_df)
combined_train_df.dropna(inplace=True)
dropped_rows = original_rows - len(combined_train_df)

if dropped_rows > 0:
    print(f"Dropped {dropped_rows} rows due to NaN values in training data.")

x_train_cleaned = combined_train_df.drop('Salary', axis=1)
y_train_cleaned = combined_train_df['Salary']


# Initialize the Linear Regression model
model = LinearRegression()

# Train the model with cleaned data
model.fit(x_train_cleaned, y_train_cleaned)

print("Linear Regression model trained successfully.")
```

### 3. Residual Plots

Residual plots display the difference between actual and predicted values (residuals) against the predicted values. A good model should show a random scattering of residuals around zero, with no discernible patterns. Patterns (e.g., a cone shape, a curve) suggest issues like heteroscedasticity or a missing non-linear relationship.

```python
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(20, 12))
axes = axes.flatten()

for i, (name, current_model) in enumerate(models.items()):
    y_pred = current_model.predict(x_test_cleaned)
    residuals = y_test_cleaned - y_pred
    ax = axes[i]
    ax.scatter(y_pred, residuals, alpha=0.6)
    ax.hlines(0, y_pred.min(), y_pred.max(), colors='r', linestyles='--', lw=2) # Zero residual line
    ax.set_title(f'{name}: Residual Plot')
    ax.set_xlabel('Predicted Salary')
    ax.set_ylabel('Residuals')
    ax.grid(True, linestyle='--', alpha=0.7)

# Hide any unused subplots
for j in range(len(models), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import r2_score, mean_absolute_error

# Prepare x_test by dropping rows with NaNs, similar to how x_train was cleaned
# We'll re-create combined_test_df to ensure consistency
combined_test_df = pd.concat([x_test, y_test], axis=1)
original_test_rows = len(combined_test_df)
combined_test_df.dropna(inplace=True)
dropped_test_rows = original_test_rows - len(combined_test_df)

if dropped_test_rows > 0:
    print(f"Dropped {dropped_test_rows} rows from test data due to NaN values.")

x_test_cleaned = combined_test_df.drop('Salary', axis=1)
y_test_cleaned = combined_test_df['Salary']


models = {
    'Linear Regression': model, # 'model' is the Linear Regression model trained earlier
    'Random Forest': rf_model,
    'Ridge': ridge_model,
    'Lasso': lasso_model,
    'Gradient Boosting': gb_model
}

performance = {}

for name, current_model in models.items():
    y_pred = current_model.predict(x_test_cleaned)
    r2 = r2_score(y_test_cleaned, y_pred)
    mae = mean_absolute_error(y_test_cleaned, y_pred)
    performance[name] = {'R2 Score': r2, 'MAE': mae}

performance_df = pd.DataFrame.from_dict(performance, orient='index')
print("\nModel Performance on Test Set:")
display(performance_df)
```

```python
best_model_name = performance_df['R2 Score'].idxmax()
best_model = models[best_model_name]

print(f"The best model based on R-squared is: {best_model_name}")

import joblib

# Define a filename for saving the model
filename = f'{best_model_name.replace(" ", "_").lower()}_model.joblib'

# Save the best model to a file
joblib.dump(best_model, filename)

print(f"Best model '{best_model_name}' saved as '{filename}'")
```
