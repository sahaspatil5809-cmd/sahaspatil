```python
import pandas as pd
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Salary%20Data.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
x=df["Experience Years"]
y=df["Salary"]
plt.scatter(x,y)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=2529)
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(x_train.to_frame(),y_train)
```

```python
lr.score(x_train.to_frame(),y_train)
```

```python
year=float(input("enter your fear o experience:"))
print(lr.predict([[year]]))
```

```python
import numpy as np
lr.intercept_
np.float64(24122.66886236802)
```

```python
lr.coef_
np.array([9750.7957687])
```

```python
import matplotlib.pyplot as plt
x=df["Experience Years"]
y=df["Salary"]
plt.figure(figsize=(10,3))
plt.scatter(x,y)
plt.plot(x,lr.predict(x.to_frame()),color='red')
plt.show()
```
